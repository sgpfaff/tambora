# Overview of Force Classes

All forces in tambora are descendents of the {class}`~tambora.dynamics.forces.Force` abstract base class. As enforced by {class}`~tambora.dynamics.forces.Force`, all forces have an {method}`~tambora.dynamics.forces.Force.acc()` method.

The acceleration a particle feels can depend on many factors, such as:
* **its own** position, velocity, and/or mass
* the positions, velocities, and/or mass of **other particles**
* the **time** it is evaluated

Which of these factors determines how a particle accelerates depends on the force that produces it. Different forces can require various combinations of these properties depending on the physical mechanism it encodes. 

Thus, descendents of {class}`tambore.dynamics.forces.Force` implement their own `acc()` method with only the arguments needed to compute its acceleration. The rest of this page describes the existing {class}`tambore.dynamics.forces.Force` subclasses and their unique qualities.

## Conservative Forces

Some forces possess additional mathematical properties of interest. Namely, **conservative** forces are special in that they can be implicitly defined by a potential. Forces are identified as conservative in tambora with the {class}`~tambora.dynamics.forces.Conservative` mixin that provides additional methods for the potential, namely the `potential()` method.

In some cases, it is more efficient to calculate the acceleration and potential together (e.g., in direct summation). Thus, there is also an optional `acc_and_potential` method for conservative forces to provide an efficient way to calculate both acceleration and potential if available. By default, `acc_and_potential` calculates the acceleration and potential separately and returns them together. Forces/Force solvers that *can* calculate both together faster than separately override the default with its own `acc_and_potential` method.

## External Forces

For simplicity, **external forces** are those that are constructed and act on particles individually. That is to say, external forces act on particles with no knowledge of the other particles. Broadly speaking, these forces encode interaction mechanisms between particles and their (often idealized) environment.

## Custom Forces

If you'd like to add a custom force to your simulation or want an easy way to combine it with other forces in tambora, you can create a new subclass of one of the existing force classes or create your own for your force. The input and output of the methods must be in [tambora internal units](units.ipynb#units-conventions) for it to be combined with other forces and integrated into a simulation properly.
